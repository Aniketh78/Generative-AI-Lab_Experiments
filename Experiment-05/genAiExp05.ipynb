```python
import google.generativeai as genai
from datasets import load_dataset

genai.configure(api_key="this-isnottheapikeyokayokayofcnotweknowitisnot")

dataset = load_dataset("gsm8k", "main", split="test")
sample = dataset[0]

question = sample['question']
answer = sample['answer']

print("Question:", question)

model = genai.GenerativeModel("gemini-3-flash-preview")

zero_prompt = f"Q: {question}\nA:"
zero_output = model.generate_content(zero_prompt).text.strip()

few_prompt = f"""
Q: If 2 pencils cost 10 rupees, what is the cost of 4 pencils?
A: 20

Q: If there are 6 bags and each has 3 books, how many books are there?
A: 18

Q: {question}
A:
"""
few_output = model.generate_content(few_prompt).text.strip()

cot_prompt = f"Q: {question}\nA: Let's think step by step."
cot_output = model.generate_content(cot_prompt).text.strip()

print("\n--- RESULTS ---")
print("Actual Answer:", answer)
print("Zero-shot:", zero_output)
print("Few-shot:", few_output)
print("Chain-of-Thought:", cot_output)
```
